```python
import pandas as pd
import numpy as np

# 1. CSV file එක Pandas DataFrame එකකට load කිරීම (Objective 2)
df = pd.read_csv("company_sales_data.csv")

# 2. මුල් පේළි කිහිපය බැලීම (Data Preview)
print("--- First 5 Rows ---")
print(df.head())

# 3. Dataset එකේ ප්‍රමාණය (Rows සහ Columns ගණන)
print("\n--- Dataset Shape (Rows, Columns) ---")
print(df.shape)

# 4. Columns වල නම්, Data Types සහ Non-Null values පරීක්ෂා කිරීම
print("\n--- Dataset Info ---")
df.info()

# 5. සංඛ්‍යාත්මක දත්ත වල Summary Statistics බැලීම
print("\n--- Summary Statistics ---")
print(df.describe())

# 6. එක් එක් Column එකේ ඇති Missing Values (හිස් තැන්) ගණන බැලීම
print("\n--- Missing Values Count ---")
print(df.isnull().sum())
```

```python
# මුල් dataset එකට හානියක් නොවෙන්න copy එකක් හදාගැනීම
df_clean = df.copy()

# 1. Duplicate rows පරීක්ෂා කර ඉවත් කිරීම
print("Duplicates before cleaning:", df_clean.duplicated().sum())
df_clean = df_clean.drop_duplicates()
print("Duplicates after cleaning:", df_clean.duplicated().sum())

# 2. Categorical (Text) columns වල Missing Values පිරවීම
df_clean["Region"] = df_clean["Region"].fillna("Unknown")
df_clean["Category"] = df_clean["Category"].fillna(df_clean["Category"].mode()[0])  # වැඩිපුරම ඇති Category එක දැමීම

# 3. Numerical columns වල Missing Values පිරවීම
# Units_Sold සඳහා Median (මධ්‍යස්ථ) අගය යෙදීම
df_clean["Units_Sold"] = df_clean["Units_Sold"].fillna(df_clean["Units_Sold"].median())

# Unit_Price සඳහා Mean (සාමාන්‍ය) අගය යෙදීම
df_clean["Unit_Price"] = df_clean["Unit_Price"].fillna(round(df_clean["Unit_Price"].mean(), 2))

# 4. Data Types නිවැරදි කිරීම සහ සංවිධානය කිරීම (Organizing Data)
df_clean["Date"] = pd.to_datetime(df_clean["Date"])
df_clean["Units_Sold"] = df_clean["Units_Sold"].astype(int)

# 5. පිරිසිදු කළ පසු Missing Values තිබේදැයි නැවත පරීක්ෂා කිරීම
print("\n--- Missing Values After Cleaning ---")
print(df_clean.isnull().sum())

print("\n--- Cleaned Dataset ---")
print(df_clean)
```

```python
# 1. අලුත් Column එකක් සෑදීම (Total_Sales = Units_Sold * Unit_Price)
df_clean["Total_Sales"] = df_clean["Units_Sold"] * df_clean["Unit_Price"]

# 2. NumPy භාවිතයෙන් සංඛ්‍යාත්මක විශ්ලේෂණය (Objective 1 & 4)
sales_array = df_clean["Total_Sales"].to_numpy()

print("--- NumPy Sales Statistics ---")
print(f"Total Revenue: Rs. {np.sum(sales_array):,.2f}")
print(f"Average Order Value (Mean): Rs. {np.mean(sales_array):,.2f}")
print(f"Median Order Value: Rs. {np.median(sales_array):,.2f}")
print(f"Standard Deviation: Rs. {np.std(sales_array):,.2f}")
print(f"Highest Single Order: Rs. {np.max(sales_array):,.2f}")

# 3. Category අනුව මුළු විකුණුම් සහ Units ගණන (Grouping)
print("\n--- Sales by Category ---")
category_summary = df_clean.groupby("Category")[["Units_Sold", "Total_Sales"]].sum()
print(category_summary)

# 4. Region අනුව මුළු විකුණුම් (Grouping & Sorting)
print("\n--- Sales by Region (Highest to Lowest) ---")
region_summary = df_clean.groupby("Region")["Total_Sales"].sum().sort_values(ascending=False)
print(region_summary)

# 5. Filtering: Rs. 100,000 ට වඩා වැඩි විකුණුම් (High-Value Orders) තෝරා ගැනීම
print("\n--- High-Value Orders (> Rs. 100,000) ---")
high_value_orders = df_clean[df_clean["Total_Sales"] > 100000].sort_values(by="Total_Sales", ascending=False)
print(high_value_orders[["Order_ID", "Date", "Region", "Category", "Total_Sales"]])

# 6. පිරිසිදු කළ dataset එක අලුත් CSV file එකක් ලෙස save කිරීම
df_clean.to_csv("cleaned_company_sales_data.csv", index=False)
print("\nCleaned dataset saved as 'cleaned_company_sales_data.csv'!")
```
