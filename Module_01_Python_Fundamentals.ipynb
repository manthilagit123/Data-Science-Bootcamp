```python
# 1. Variables & Data Types (දත්ත ගබඩා කිරීම)
project_name = "Data Science Bootcamp"  # String (අකුරු)
total_days = 4                          # Integer (පූර්ණ සංඛ්‍යා)
completion_rate = 25.5                  # Float (දශම සංඛ්‍යා)
is_active = True                        # Boolean (සත්‍ය/අසත්‍ය)

print("Project:", project_name)
print("Progress:", completion_rate, "%\n")

# 2. Lists (එකම වර්ගයේ හෝ වෙනස් දත්ත ලැයිස්තුවක්)
daily_scores = [85, 92, 78, 95]
print("All Scores:", daily_scores)
print("First Day Score:", daily_scores[0])  # Python වල ගණන් කිරීම පටන් ගන්නේ 0 න්
daily_scores.append(88)                     # අලුත් අගයක් අගට එකතු කිරීම
print("Updated Scores:", daily_scores, "\n")

# 3. Dictionaries (Key සහ Value යුගල වශයෙන් දත්ත තබා ගැනීම - Data Science වල බහුලව භාවිතා වේ)
student_profile = {
    "name": "Manthila",
    "module": "Module 01",
    "topics_completed": ["Setup", "Variables", "Lists"],
    "average_score": 87.6
}

print("Student Name:", student_profile["name"])
print("Completed Topics:", student_profile["topics_completed"])

# Dictionary එකකට අලුත් Key-Value එකක් එකතු කිරීම
student_profile["status"] = "In Progress"
print("Updated Profile:", student_profile)
```

```python
# 1. Loops & Conditions (දත්ත පරීක්ෂා කිරීම සහ තෝරා ගැනීම)
scores = [85, 92, 78, 95, 88, 64, 91]

print("--- High Scores (85 or above) ---")
for score in scores:
    if score >= 85:
        print(f"Score {score}: Excellent!")
    else:
        print(f"Score {score}: Needs Improvement")

# 2. Functions (නැවත නැවත පාවිච්චි කළ හැකි කේත කොටස් සෑදීම)
def analyze_data(data_list):
    total = sum(data_list)
    count = len(data_list)
    average = total / count
    highest = max(data_list)
    lowest = min(data_list)

    # ප්‍රතිඵලය Dictionary එකක් ලෙස ආපසු ලබා දීම
    return {
        "Total Items": count,
        "Average": round(average, 2),
        "Highest": highest,
        "Lowest": lowest
    }

# අපි හදපු Function එක call (ක්‍රියාත්මක) කිරීම
summary = analyze_data(scores)
print("\n--- Data Summary ---")
for key, value in summary.items():
    print(f"{key}: {value}")
```

## Objective 4: Data Science Concepts & Real-World Applications

### 1. What is Data Science?
Data Science is an interdisciplinary field that combines programming (Python), statistics, and domain knowledge to extract meaningful insights and predict future trends from structured and unstructured data.

### 2. Core Workflow of a Data Science Project
* **Data Collection:** Gathering raw data from files (CSV, JSON), databases, or APIs.
* **Data Cleaning:** Handling missing values, removing duplicates, and fixing formatting errors.
* **Exploratory Data Analysis (EDA):** Analyzing patterns, averages, and correlations using Python.
* **Modeling & Prediction:** Using Machine Learning algorithms to forecast future outcomes.

### 3. Real-World Applications
* **Construction & Engineering Management:** Predicting project cost overruns, optimizing material procurement (e.g., steel, concrete, MEP components), and forecasting project completion timelines using historical site data.
* **E-Commerce & Retail:** Recommendation engines (e.g., Amazon, Daraz) that suggest products based on user browsing and purchase history.
* **Banking & Finance:** Real-time fraud detection systems that identify unusual credit card transactions and credit risk scoring.
* **Healthcare:** Early disease detection from medical imaging and predicting patient admission rates.
